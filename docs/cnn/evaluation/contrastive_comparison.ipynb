# Contrastive retrieval comparison

The default comparison is MLP versus the mixed-baseline CNN on identical domain subsets. Set `INCLUDE_FINETUNED=True` only for an explicit ablation.

```python
import pandas as pd
from neurovlm import AtlasFreeCNNDataProvider
from neurovlm.evaluation import (
    default_comparison_matrix, evaluate_contrastive_comparison,
)

DOMAINS = ("pubmed", "nilearn", "neurovault")
LIMIT_PER_DOMAIN = 32
INCLUDE_FINETUNED = False
```

```python
results = []
for domain in DOMAINS:
    selections = default_comparison_matrix(
        "contrastive", domains=(domain,), include_finetuned=INCLUDE_FINETUNED
    )
    results.append(evaluate_contrastive_comparison(
        selections=selections,
        provider=AtlasFreeCNNDataProvider(domain=domain, limit=LIMIT_PER_DOMAIN),
    ))
summary = pd.DataFrame(row for result in results for row in result.summary)
summary
```

## Preserved executed ablation evidence

The historical run used 32 held-out examples per domain. Its mean normalized recall-curve AUC was:

| domain | mixed baseline | fine-tuned |
|---|---:|---:|
| PubMed | 0.807617 | 0.804688 |
| Nilearn | 0.907227 | 0.907227 |
| NeuroVault | 0.850586 | 0.845703 |

This small, limit-32 comparison supports the mixed baseline as the product default; it is not a claim of universal superiority. The complete preserved CSV, JSON, and PNG evidence is under `artifacts/contrastive_retrieval/`.
